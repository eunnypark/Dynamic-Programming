```python
# 피보나치수열 재귀 시간복잡도 체감

def f(x):
  if x==0 or x==1:
    return 1
  return f(x-1)+f(x-2)

# print(f(10))  89. 결과 바로 나옴.
# print(f(20))  10946. 결과 바로 나옴.
# print(f(30))  1346269. 결과 바로 나옴. (1초 이하의 딜레이)
# print(f(35))  14930352. 결과 3초 딜레이
# print(f(40))  165580141. 결과 30초 딜레이
# print(f(50))  13분 동안 결과가 나오지 않아 중단. 초당 수백만 회의 함수 호출을 처리한다 가정했을 때 407억회 재귀호출에는 수 시간이 걸림.

# 시간복잡도 2^n
```

```python
# 에어텔 문제 (Airtel Problem) 구현

n = 6
A = [0, 1, 3, 6, 11, 17]  # A[i]: 도시 간 거리가 i일 때의 항공요금
H = [0, 2, 5, 1, 5, 0]  # H[k]: 도시 k에서의 호텔 숙박비


# 1. 분할통치법 (정방향) 구현 (41p 의사코드 기반)
def airtel_divide_and_conquer(n, A, H):
    def rAirtel(d):

        if d == 0:
            return 0

        mincost = float('inf')

        for k in range(d):  # k는 경유 도시 (0부터 d-1까지)
            cost = rAirtel(k) + H[k] + A[d - k]
            if cost < mincost:
                mincost = cost
        return mincost

    return rAirtel(n - 1)


# 2. 동적프로그래밍 (정방향) 구현 (44p 의사코드 기반)
def airtel_dynamic_programming(n, A, H):

    m = [0] * n # 중간 계산값 저장할 크기 n의 배열 m
    m[0] = 0  # 출발도시 0에 대한 최소비용 초기화

    for d in range(1, n): # 도착도시 d를 1부터 n-1까지 순차적으로 진행
        m[d] = float('inf')
        for k in range(d):  # 경유 도시 k를 0부터 d-1까지 순회하며 최소 비용 계산
            cost = m[k] + H[k] + A[d - k]
            if cost < m[d]:
                m[d] = cost

    return m[n - 1]


# print
print(f"분할통치법: {airtel_divide_and_conquer(n, A, H)}")
print(f"동적프로그래밍: {airtel_dynamic_programming(n, A, H)}")
```
